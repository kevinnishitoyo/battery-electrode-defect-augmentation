```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Load the original labels
labels_df = pd.read_csv(
    "../../data/raw/archive/classification/labels.csv"
)

# Defect columns
defect_columns = [
    "Surface_Crack",
    "Delamination",
    "Pinhole",
    "unclassified"
]

# Count how many labels each image has
labels_df["number_of_labels"] = labels_df[defect_columns].sum(axis=1)

# Keep only images with exactly one label
single_label_df = labels_df[
    labels_df["number_of_labels"] == 1
].copy()

# Create one class column
single_label_df["class_name"] = (
    single_label_df[defect_columns].idxmax(axis=1)
)

# Remove unclassified images
model_df = single_label_df[
    single_label_df["class_name"] != "unclassified"
].copy()

print(model_df["class_name"].value_counts())
```

```python
train_df, temp_df = train_test_split(
    model_df,
    test_size=0.3,
    random_state=42,
    stratify=model_df["class_name"]
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.5,
    random_state=42,
    stratify=temp_df["class_name"]
)
```

```python
train_df.to_csv("../data/processed/train_split.csv", index=False)
val_df.to_csv("../data/processed/val_split.csv", index=False)
test_df.to_csv("../data/processed/test_split.csv", index=False)
```
